```python
import numpy as np
import pandas as pd

grid=pd.read_csv('grid_name.csv',header=None)
```

```python
# print(grid)
grid=grid.to_numpy(na_value='   ')
# print(grid[10,4])
```

```python
id=pd.read_csv('id.csv',usecols=[0,1],dtype='str')
id.set_index(id['collegi'],inplace=True)
id.drop('collegi',axis=1, inplace=True)
# print(id.head(10))
id=id.to_dict('index')
```

```python

i,j=grid.shape

for k in range(0,i) :
    for l in range(0,j):
        for m in id.keys():
            if grid[k,l]==m:
                grid[k,l]=id[m]['id']
                # print(grid[k,l])
```

```python
df=pd.DataFrame(grid)
# print(df)
df.to_csv('grid.csv',header=False, index=False)
```
